## Part 3 — Boolean Indexing

Create:
```python

```
1.	Extract all values greater than 15.

````python
x = np.array([[12, 5, 18, 3],
              [25, 7, 14, 9],
              [2, 30, 11, 20]])

print(x[x > 15])
print(x[x <= 10])

x_modified = x.copy()
x_modified[x_modified > 20] = 0
print(x_modified)

print(np.sum(x > 15))
````

## Part 5 — Matrix Multiplication
Create:
```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8],
              [9, 10],
              [11, 12]])
```
1.	Compute A @ B.
2.	Compute the same result using np.dot(A, B).
3.	Verify that both results are identical.
4.	Print the shape of A, B, and the result.

The notebook demonstrates the equivalence of .dot(), np.dot(), and @ for matrix multiplication.

````python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8],
              [9, 10],
              [11, 12]])

print(A @ B)
print(np.dot(A, B))
print(np.array_equal(A @ B, np.dot(A, B)))
print(f"A: {A.shape}, B: {B.shape}, Result: {(A @ B).shape}")
````

## Part 6 — Transpose and Axis Operations
Using:
```python
X = np.array([[2, 4, 6],
              [8, 10, 12]])
```
Calculate:
1.	X.T
2.	np.transpose(X)
3.	np.sum(X)
4.	np.sum(X, axis=0)
5.	np.sum(X, axis=1)
6.	np.mean(X)
7.	np.mean(X, axis=0)
8.	np.mean(X, axis=1)
9.	np.max(X, axis=0)
10.	np.min(X, axis=1)
11.	np.std(X)

For each axis operation, write one sentence explaining what the axis represents.

````python
X = np.array([[2, 4, 6],
              [8, 10, 12]])

print(X.T)
print(np.transpose(X))
print(np.sum(X))
print(np.sum(X, axis=0)) # axis=0 represents operations down the columns (vertically across rows).
print(np.sum(X, axis=1)) # axis=1 represents operations across the rows (horizontally across columns)
print(np.mean(X))
print(np.mean(X, axis=0)) # Calculates mean for each column across rows.
print(np.mean(X, axis=1)) # Calculates mean for each row across columns.
print(np.max(X, axis=0)) # Finds maximum element in each column.
print(np.min(X, axis=1)) # Finds minimum element in each row.
print(np.std(X))
````

## Part 7 — Mini Application: Student Grades

Create a NumPy array containing the grades of 5 students in 4 assignments:

* Student 1: 85, 90, 78, 92
* Student 2: 70, 88, 91, 76
* Student 3: 95, 93, 89, 97
* Student 4: 60, 72, 68, 75
* Student 5: 88, 84, 90, 86

Using NumPy only:
1.	Calculate each student’s average.
2.	Find the highest grade on each assignment.
3.	Find the average grade for each assignment.
4.	Find the student with the highest overall average.
5.	Find all grades below 70.
6.	Replace grades below 70 with 70.
7.	Calculate the overall class average.

````python
grades = np.array([
    [85, 90, 78, 92], 
    [70, 88, 91, 76], 
    [95, 93, 89, 97], 
    [60, 72, 68, 75], 
    [88, 84, 90, 86] 
])

print(np.mean(grades, axis=1))
print(np.max(grades, axis=0))
print(np.mean(grades, axis=0))
print(np.argmax(np.mean(grades, axis=1)))
print(grades[grades<70])
grades_updated = grades.copy()
grades_updated[grades_updated < 70] = 70
print(grades_updated)
print(np.mean(grades))
````
