Part 9 — Automatic Differentiation

PyTorch provides automatic differentiation using autograd.

````python
x = torch.tensor([[1., 2.],
                  [3., 4.]], requires_grad=True)

# TODO: Calculate y = 3x² + 2
y = 3 * (x ** 2) + 2

````

````python
# Then calculate:
z = torch.sum(y)

# Use:
z.backward()
# to calculate: dz/dx

# Print:

print(y)
print(z)
print(x.grad)
````

Questions
* What does requires_grad=True mean?
* What does backward() do?
* What is stored in x.grad?
* Why is automatic differentiation important for neural networks?

1. It tells Pytorch to track all computational operations performed on this tensor so that gradients can be automatically calculated during backpropagation.
2. It initiates the backpropagation process, computing the partial derivatives of the scalar output z.
3. It stores the calculated gradient values dz/dx.
4. It allows model weights and parameters to update automatically without manual gradient derivation.
