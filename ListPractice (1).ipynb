#### Problem 1:

Write a program that stores the names of three of your favorite movies in a list called movies. Then:

*	Print the entire list
*	Print the first and last movie in the list
*	Change the second movie to a new title of your choice
*	Print the updated list

**Sample output**
```
['Forrest Gump', 'Good Will Hunting', 'The Big Short']
First movie: Forrest Gump
Last movie: The Big Short
Updated list: ['Forrest Gump', 'Princess Bride', 'The Big Short']

```

````python
# TODO: create the list of movies
# Step 1: Print the list
# Step 2: Print the first and last movies
# Step 3: Change the second movie
# Step 4: Print the updated list
movies = ['Forrest Gump', 'Good Will Hunting', 'The Big Short']
print(movies)
print(f"First movie: {movies[0]}")
print(f"Last movie: {movies[-1]}")
movies[1] = 'Princess Bride'
print(f"Updated list: {movies}")
````

# Problem 2
You have a list of friends attending a dinner party: Farah, Alex, Charlie, Dana, Brooke, Eli.

Create the list, then:

1. Use slicing to print only the middle guests (Charlie, Dana, Eli).
2. Add a new friend "Gina" to the end of the list.
3. Remove "Alex" from the list.
4. Print the final sorted list in alphabetical order.

**Sample output**
```
Middle guests: ['Charlie', 'Dana', 'Brooke']
Final guest list: ['Brooke', 'Charlie', 'Dana', 'Eli', 'Farah', 'Gina']
```


````python
# TODO: Create the friends list

# Step 1: Slice to get middle guests
# Step 2: Append a new friend
# Step 3: Remove a friend
# Step 4: Sort and print the final list
friends = ['Charlie', 'Dana', 'Brooke', 'Eli', 'Farah', 'Alex']
print(f"Middle guests: {friends[0:3]}")
friends.append("Gina")
friends.remove("Alex")
friends.sort()
print(f"Final guest list: {friends}")
````

# Problem 3
You have two lists of team members:
- Team A: Liam, Olivia, Noah
- Team B: Emma, Ava, Sophia

Create one list for each team, then:
1. Combine both lists into a new list called all_members and print this new list.
2. Add "Mason" to the combined list.
3. Remove "Olivia" from the list.
4. Print the total number of members using the `len()` function.
5. Print the sorted list of all members.

**Sample output**
```
['Liam', 'Olivia', 'Noah', 'Emma', 'Ava', 'Sophia']
Total team members: 6
['Ava', 'Emma', 'Liam', 'Mason', 'Noah', 'Sophia']
```

````python
# TODO: create both lists

# Step 1: Combine lists and print them
# Step 2: Add a new member
# Step 3: Remove a member
# Step 4: Count and print total members
# Step 5: Sort and print final list
teamA = ['Liam', 'Olivia', 'Noah']
teamB = ['Emma', 'Ava', 'Sophia']

all_members = teamA + teamB
print(all_members)
all_members.append("Mason")
all_members.remove("Olivia")
print(f"Total team members: {len(all_members)}")
all_members.sort()
print(all_members)
````

### Problem 4:

Given a list of ['cat', 'mouse', 'cheeta', 'zebra'], sort the list on the string length (same as in the lecture slide)


````python
# TODO:
# Declare the list
# Step 1: Print the list
# Step 2: declare a function to return the length of a string
# Step 3: sort the list based on the string length
# Step 4: Print the sorted list
animals = ['cat', 'mouse', 'cheeta', 'zebra']
print(animals)

def length(s):
    return len(s)

animals.sort(key=length)
print(animals)
````
