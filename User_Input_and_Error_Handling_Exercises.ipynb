## Problem 1

Write a bill-splitting program that:
1. Asks the user for the total bill amount
2. Asks the user for the number of people splitting the bill
3. Prints how much each person owes, rounded to 2 decimal places

Make sure to cast the inputs to the appropriate numeric types.

**Sample output**
```
Each person owes $15.00.
```

````python
# TODO write your program here
total_bill = float(input("Enter total bill amount: "))
people_count = int(input("Enter number of people splitting: "))

per_person = total_bill / people_count

print(f"Each person owes ${per_person:.2f}.")
````

## Problem 2

Update your bill-splitting program to handle the case where the user enters something that isn't a number. Use a `try/except` block to catch this error and print: `"Please enter valid numbers."`

**Sample output**
```
Each person owes $15.00.
```
```
Please enter valid numbers.
```

````python
# TODO write your program here
try:
    total_bill = float(input("Enter total bill amount: "))
    people_count = int(input("Enter number of people splitting: "))

    per_person = total_bill / people_count

    print(f"Each person owes ${per_person:.2f}.")
except ValueError:
    print("Please enter valid numbers.")
````

## Problem 3

Update your bill-splitting program one more time to also handle the case where the user enters `0` for the number of people. Use two `except` blocks to handle these two error cases separately:
- If either input is not a number, print: `"Please enter valid numbers."`
- If the number of people is zero, print: `"Error: The number of people cannot be zero."`

**Sample output**
```
Each person owes $15.00.
```
```
Please enter valid numbers.
```
```
Error: The number of people cannot be zero.
```

````python
# TODO write your program here
try:
    total_bill = float(input("Enter total bill amount: "))
    people_count = int(input("Enter number of people splitting: "))

    per_person = total_bill / people_count

    print(f"Each person owes ${per_person:.2f}.")
except ValueError:
    print("Please enter valid numbers.")
except ZeroDivisionError:
    print("Error: The number of people cannot be zero.")
````
